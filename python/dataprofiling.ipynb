````python
```python
import pandas as pd
import os

# =========================================================
# 1. DATASET LOCATION
# =========================================================

folder = "datasets"

# =========================================================
# 2. DATASETS
# =========================================================

files = {
    "Renewable": "Renewable Energy monitoring.xlsx",
    "SCADA": "scada_pipeline.csv",
    "Smart Meter": "smart_meter_data.csv",
    "Maintenance": "asset_maintenance.csv",
    "Billing": "Billing_payments.csv",
    "IIoT": "iiot_smart_grid_dataset.csv",
    "OWID": "owid-energy-data.json",
    "OMS": "OMS_data.csv"
}

# =========================================================
# 3. LOAD DATASETS
# =========================================================

def load_file(filename):
    path = os.path.join(folder, filename)

    if filename.endswith(".csv"):
        return pd.read_csv(path)
    elif filename.endswith(".xlsx"):
        return pd.read_excel(path, sheet_name="data")
    elif filename.endswith(".json"):
        return pd.read_json(path)

data = {}

for name, file in files.items():
    try:
        data[name] = load_file(file)
        print(f"{name} loaded successfully")
    except Exception as e:
        print(f"Error loading {name}: {e}")

# =========================================================
# 4. DATA PROFILING
# =========================================================

profile = []

for dataset, df in data.items():

    print("\n" + "=" * 70)
    print(dataset)
    print("=" * 70)

    print("Rows:", len(df))
    print("Columns:", len(df.columns))

    for column in df.columns:

        s = df[column]

        missing = s.isna().sum()
        missing_pct = round(missing / len(s) * 100, 2)
        unique = s.nunique()
        duplicates = s.duplicated().sum()

        minimum = ""
        maximum = ""
        mean = ""

        if pd.api.types.is_numeric_dtype(s):
            minimum = s.min()
            maximum = s.max()
            mean = round(s.mean(), 2)

        print(f"\nColumn: {column}")
        print(f"Data Type: {s.dtype}")
        print(f"Missing Values: {missing}")
        print(f"Missing %: {missing_pct}%")
        print(f"Unique Values: {unique}")
        print(f"Duplicate Values: {duplicates}")

        if minimum != "":
            print(f"Minimum: {minimum}")
            print(f"Maximum: {maximum}")
            print(f"Mean: {mean}")

        print("Sample Values:",
              s.dropna().head(3).tolist())

        profile.append([
            dataset,
            column,
            str(s.dtype),
            len(s),
            missing,
            missing_pct,
            unique,
            duplicates,
            minimum,
            maximum,
            mean
        ])

# =========================================================
# 5. CREATE PROFILING REPORT
# =========================================================

profile_df = pd.DataFrame(
    profile,
    columns=[
        "Dataset",
        "Column",
        "Data Type",
        "Records",
        "Missing Values",
        "Missing %",
        "Unique Values",
        "Duplicate Values",
        "Minimum",
        "Maximum",
        "Mean"
    ]
)
profile_df.to_excel(
    "Data_Profiling_Report.xlsx",
    index=False
)
print("\nData Profiling completed.")
print("Created: Data_Profiling_Report.xlsx")
```
````
